```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "Helsinki-NLP/opus-mt-en-hi"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

text = "I'm learning transformers in Google Colab"
inputs = tokenizer(text, return_tensors="pt")
outputs = model.generate(**inputs, max_new_tokens=100)

print(tokenizer.batch_decode(outputs, skip_special_tokens=True)[0])
```

# I want to classify text into a category
Input: 'India won world cup in Circket'
I want you to find a classifier which classifies the input and possible options are '['Sports', 'Politics', 'Technology']'

```python
classifier = pipeline(
    "text-classification",
    model="distilbert/distilbert-base-uncased-finetuned-sst-2-english")

resp = classifier("India won world cup in Circket")
print(resp)


```

```python
classifier = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")

sequence_to_clarity = "India won world cup in Circket"
input_options = ['Sports', 'Politics', 'Technology', 'health']

result = classifier(sequence_to_clarity, input_options)
print(result)
```

```python
from transformers import AutoModelForCausalLM, AutoTokenizer

model_name = "TinyLlama/TinyLlama-1.1B-Chat-v1.0"
# Load tokenizer and model
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

# Tokenize input and generate text
inputs = tokenizer("Inertia in physics is all about", return_tensors="pt")
outputs = model.generate(**inputs, max_new_tokens=50)
print(tokenizer.decode(outputs[0], skip_special_tokens=True))
```
